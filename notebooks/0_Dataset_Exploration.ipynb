```python
import pandas as pd

xls = pd.ExcelFile(
    r"C:\BRCA\BRCA Project\data\integratedatlas_supplemental_files\SuppTable1.xlsx"
)

print(xls.sheet_names)
```

```python
for sheet in xls.sheet_names:

    df = pd.read_excel(xls, sheet_name=sheet)

    print("=" * 80)
    print(sheet)

    missing = (
        df.isna()
        .sum()
        .sort_values(ascending=False)
    )

    print(missing[missing > 0])
```
